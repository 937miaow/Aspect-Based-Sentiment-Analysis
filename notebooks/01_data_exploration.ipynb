## 导入依赖

```python
import sys
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sys.path.append(os.path.abspath('..'))
from src.utils.dataloader import load_semeval_xml
```

## 加载数据

```python
train_path = "../data/raw/semeval2014_task4/Restaurants_Train.xml"
df = load_semeval_xml(train_path)
print(f"样本数量: {len(df)}")
df.head()
```

## 数据可视化 - 情感分布

```python
all_aspects = []
for aspects in df['gold_aspects']:
    all_aspects.extend(aspects)
    
aspect_df = pd.DataFrame(all_aspects)

# 绘制情感分布
sns.countplot(x='sentiment', data=aspect_df)
plt.title('Sentiment Distribution in Training Set')
plt.show()
```

## 数据预处理与保存

```python
grouped_df = df.groupby('text').apply(
    lambda x: x[['aspect_term', 'sentiment']].to_dict('records')
).reset_index().rename(columns={0: 'aspects'})

output_path = "../data/processed/semeval_restaurants_train.jsonl"
os.makedirs(os.path.dirname(output_path), exist_ok=True)
grouped_df.to_json(output_path, orient='records', lines=True)

print(f"已将处理后的数据保存至: {output_path}")
display(grouped_df.head(2))
```
